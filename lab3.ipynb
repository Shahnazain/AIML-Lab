```python
#A* ALGORITHM
# Graph: each node has its neighboring nodes and the cost to reach them
graph = {
    'A': {'B': 1, 'C': 3},
    'B': {'D': 3, 'E': 1},
    'C': {'F': 2},
    'D': {'G': 2},
    'E': {'G': 2},
    'F': {'G': 1},
    'G': {}
}

# Heuristic values (estimated cost from each node to goal G)
h = {
    'A': 5,
    'B': 4,
    'C': 3,
    'D': 2,
    'E': 2,
    'F': 1,
    'G': 0
}


def a_star(start, goal):
    # Nodes that we still need to explore
    open_list = [start]

    # Cost from start to each node
    g_cost = {start: 0}

    # To store the previous node
    parent = {start: None}

    while open_list:
        # Select node with the smallest f = g + h
        current = min(open_list, key=lambda node: g_cost[node] + h[node])

        # If goal is reached
        if current == goal:
            path = []

            # Build the path backwards
            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1], g_cost[goal]

        # Remove current node from open list
        open_list.remove(current)

        # Check all neighbors
        for neighbor, cost in graph[current].items():

            # Calculate new cost
            new_cost = g_cost[current] + cost

            # If this is a better path
            if neighbor not in g_cost or new_cost < g_cost[neighbor]:
                g_cost[neighbor] = new_cost
                parent[neighbor] = current

                if neighbor not in open_list:
                    open_list.append(neighbor)

    return None, float('inf')


# Start A and find the path to G
path, cost = a_star('A', 'G')

print("Shortest Path:", path)
print("Total Cost:", cost)
```

```python
# Uniform Cost Search (UCS)

# Graph: each node has its neighboring nodes and the cost
graph = {
    'A': {'B': 1, 'C': 3},
    'B': {'D': 3, 'E': 1},
    'C': {'F': 2},
    'D': {'G': 2},
    'E': {'G': 2},
    'F': {'G': 1},
    'G': {}
}


def uniform_cost_search(start, goal):

    # Nodes that we still need to explore
    open_list = [start]

    # Cost from start to each node
    cost = {start: 0}

    # To store the previous node
    parent = {start: None}

    while open_list:

        # Select the node with the lowest cost
        current = min(open_list, key=lambda node: cost[node])

        # If goal is reached
        if current == goal:

            path = []

            # Build the path backwards
            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1], cost[goal]

        # Remove current node from open list
        open_list.remove(current)

        # Check all neighbors
        for neighbor, edge_cost in graph[current].items():

            # Calculate new cost
            new_cost = cost[current] + edge_cost

            # If this is a better path
            if neighbor not in cost or new_cost < cost[neighbor]:

                cost[neighbor] = new_cost
                parent[neighbor] = current

                if neighbor not in open_list:
                    open_list.append(neighbor)

    return None, float('inf')


# Start A and find the path to G
path, total_cost = uniform_cost_search('A', 'G')

print("Shortest Path:", path)
print("Total Cost:", total_cost)
```

```python
#GREEDY BEST FIRST SEARCH ALGORITHM
# Graph: each node has its neighboring nodes
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': ['G'],
    'F': ['G'],
    'G': []
}

# Heuristic values
# Estimated distance from each node to the goal G
h = {
    'A': 5,
    'B': 4,
    'C': 3,
    'D': 2,
    'E': 2,
    'F': 1,
    'G': 0
}


def greedy_best_first(start, goal):

    # Nodes that we still need to explore
    open_list = [start]

    # To store the previous node
    parent = {start: None}

    # Keep track of visited nodes
    visited = set()

    while open_list:

        # Select node with the smallest heuristic value
        current = min(open_list, key=lambda node: h[node])

        # If goal is reached
        if current == goal:

            path = []

            # Build the path backwards
            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        # Remove current node from open list
        open_list.remove(current)

        # Mark current node as visited
        visited.add(current)

        # Check all neighbors
        for neighbor in graph[current]:

            if neighbor not in visited and neighbor not in open_list:

                parent[neighbor] = current
                open_list.append(neighbor)

    return None


# Start A and find the path to G
path = greedy_best_first('A', 'G')

print("Path:", path)
```

```python
#BFS ALGORITHM

# Graph: each node has its neighboring nodes
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': ['G'],
    'F': ['G'],
    'G': []
}


def bfs(start, goal):

    # Queue
    open_list = [start]

    # To store the previous node
    parent = {start: None}

    # Keep track of visited nodes
    visited = set()

    while open_list:

        # Take the first node from the queue
        current = open_list.pop(0)

        # Mark as visited
        visited.add(current)

        # If goal is reached
        if current == goal:

            path = []

            # Build the path backwards
            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        # Check all neighbors
        for neighbor in graph[current]:

            if neighbor not in visited and neighbor not in open_list:

                parent[neighbor] = current
                open_list.append(neighbor)

    return None


# Start A and find the path to G
path = bfs('A', 'G')

print("Path:", path)
```

```python
# Depth-First Search (DFS)

# Graph: each node has its neighboring nodes
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': ['G'],
    'F': ['G'],
    'G': []
}


def dfs(start, goal):

    # Stack
    open_list = [start]

    # To store the previous node
    parent = {start: None}

    # Keep track of visited nodes
    visited = set()

    while open_list:

        # Take the last node from the stack
        current = open_list.pop()

        # If already visited, skip it
        if current in visited:
            continue

        # Mark as visited
        visited.add(current)

        # If goal is reached
        if current == goal:

            path = []

            # Build the path backwards
            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        # Add neighbors to the stack
        for neighbor in graph[current]:

            if neighbor not in visited:
                parent[neighbor] = current
                open_list.append(neighbor)

    return None


# Start A and find the path to G
path = dfs('A', 'G')

print("Path:", path)
```
